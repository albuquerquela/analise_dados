```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
caminho_arquivo = 'seu_arquivo.xlsx'
df = pd.read_excel(caminho_arquivo, sheet_name='Exemplo Gráfico Brasil')
```

```python
TAMANHO = (18, 6)
fig, axes = plt.subplots(1, 3, figsize=TAMANHO)
cores = ['#ff9999', '#66b3ff', '#99ff99', '#ffcc99', '#c2c2f0', '#ffb3e6']
```

```python
grupos = ['Branca', 'Preta ou Parda', 'Indígena']
```

```python
for i, col in enumerate(grupos):
    axes[i].pie(
        df[col],
        labels=df['Meio de Transporte'],
        autopct='%1.3f%%',  # 3 casas decimais conforme solicitado
        startangle=180,  # Ângulo inicial de rotação
        colors=cores,
        wedgeprops={'edgecolor': 'white', 'linewidth': 1.2},
    )
    axes[i].set_title(f'Cor/Raça: {col}', fontsize=12, fontweight='bold')
```

```python
plt.suptitle(
    'Distribuição dos Meios de Transporte por Cor/Raça (Mulheres)',
    fontsize=14,
    fontweight='bold',
)
plt.tight_layout()
plt.savefig('grafico_transporte.png', dpi=300)
plt.show()
```

```python
df_br = pd.read_excel(caminho_arquivo, sheet_name='BR GR UF MU')
```

```python
if 'Nível' in df_br.columns:
    df_estados = df_br[df_br['Nível'] == 'UF'].copy()
elif 'UF' in df_br.columns:
    df_estados = df_br.dropna(subset=['UF']).copy()

estados_unicos = df_estados['UF'].unique()
print("--- LISTA DOS ESTADOS (UF) ---")
print(estados_unicos)
```

```python
colunas_identificadoras = [col for col in ['Nível', 'Região', 'UF', 'Município', 'Nome'] if col in df_br.columns]
colunas_transporte = [col for col in ['Transporte Coletivo', 'Bicicleta', 'Carro', 'A pé'] if col in df_br.columns]

df_melted = pd.melt(
    df_br,
    id_vars=colunas_identificadoras,
    value_vars=colunas_transporte,
    var_name='Meio_Transporte',
    value_value_name='Quantidade' if 'Quantidade' in df_br.columns else 'Valor'
)
```

```python
df_uf = df_br[df_br['Nível'] == 'UF'] if 'Nível' in df_br.columns else df_br
df_mu = df_br[df_br['Nível'] == 'MU'] if 'Nível' in df_br.columns else df_br
```

```python
estado_mais_tc = df_uf.groupby('UF')['Transporte Coletivo'].sum().idxmax()
val_mais_tc = df_uf.groupby('UF')['Transporte Coletivo'].sum().max()
```

```python
estado_menos_tc = df_uf.groupby('UF')['Transporte Coletivo'].sum().idxmin()
val_menos_tc = df_uf.groupby('UF')['Transporte Coletivo'].sum().min()
```

```python
col_cidade = 'Município' if 'Município' in df_mu.columns else 'Nome'
cidade_mais_bike = df_mu.groupby(col_cidade)['Bicicleta'].sum().idxmax()
val_mais_bike = df_mu.groupby(col_cidade)['Bicicleta'].sum().max()
```

```python
cidade_mais_carro = df_mu.groupby(col_cidade)['Carro'].sum().idxmax()
val_mais_carro = df_mu.groupby(col_cidade)['Carro'].sum().max()
```

```python
print("\n" + "="*50)
print("RESPOSTAS ÀS QUESTÕES DE MOBILIDADE")
print("="*50)
print(f"1. Estado com MAIS mulheres no transporte coletivo: {estado_mais_tc} ({val_mais_tc:,.0f})")
print(f"2. Estado com MENOS mulheres no transporte coletivo: {estado_menos_tc} ({val_menos_tc:,.0f})")
print(f"3. Cidade com MAIS mulheres de bicicleta: {cidade_mais_bike} ({val_mais_bike:,.0f})")
print(f"4. Cidade com MAIS mulheres de carro: {cidade_mais_carro} ({val_mais_carro:,.0f})")
```
